```python
from langchain_core.tools import tool
def multiply(a,b):
    '''Multiply two numbers'''
    return a*b
```

```python
def multiply(a:int,b:int)->int:
    '''Multiplty two numbers'''
    return a*b
```

```python
@tool
def multiply(a:int,b:int)->int:
    '''Multiplty two numbers'''
    return a*b
```

```python
result = multiply.invoke({'a':5,'b':3})
print(result)
print(multiply.args_schema.model_json_schema())
```

### M:2 - Using Structured Tool 

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field

class MultiplyInput(BaseModel):
    a:int = Field(required=True,description="first integer for multiplication")
    b:int = Field(required=True,description="second integer for multiplication")

def multiply_func(a:int,b:int)->int:
    return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiply Two Numbers",
    args_schema=MultiplyInput
)
result=multiply_tool.invoke({'a':5,'b':4})
print(result)
```

### M-3 - ToolKit

```python
from langchain_core.tools import tool
@tool
def multiply(a:int,b:int)->int:
    '''Multiplty two numbers'''
    return a*b
```

```python
@tool
def add(a:int,b:int)->int:
    '''Addition two numbers'''
    return a+b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
